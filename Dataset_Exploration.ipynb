```python

import pandas as pd

# Step 1: Create the dataset
data = {
    "StudentID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Anu", "Bala", "Chitra", "Deepa", "Ezhil",
             "Farhan", "Gokul", "Hari", "Indhu", "Jaya"],
    "Gender": ["Female", "Male", "Female", "Female", "Male",
               "Male", "Male", "Male", "Female", "Female"],
    "Attendance": [95, 80, 90, 75, 85, 70, 92, 60, 88, 78],
    "Marks": [88, 72, 95, 65, 82, 58, 91, 55, 86, 76]
}

# Step 2: Convert data into a DataFrame
df = pd.DataFrame(data)

# Step 3: Display first five records
print("FIRST FIVE RECORDS")
print(df.head())

# Step 4: Display last five records
print("\nLAST FIVE RECORDS")
print(df.tail())

# Step 5: Display rows and columns
print("\nDATASET SHAPE")
print(df.shape)

# Step 6: Display column names
print("\nCOLUMN NAMES")
print(df.columns.tolist())

# Step 7: Display dataset information
print("\nDATASET INFORMATION")
df.info()

# Step 8: Display statistical summary
print("\nSTATISTICAL SUMMARY")
print(df.describe())

# Step 9: Identify numerical columns
print("\nNUMERICAL COLUMNS")
print(df.select_dtypes(include=["number"]).columns.tolist())

# Step 10: Identify categorical columns
print("\nCATEGORICAL COLUMNS")
print(df.select_dtypes(include=["object"]).columns.tolist())

# Step 11: Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# Step 12: Five observations
print("\nOBSERVATIONS")
print("1. Number of rows:", df.shape[0])
print("2. Number of columns:", df.shape[1])
print("3. Average marks:", round(df["Marks"].mean(), 2))
print("4. Highest marks:", df["Marks"].max())
print("5. Lowest marks:", df["Marks"].min())
```
